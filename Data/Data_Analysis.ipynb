```python
import pandas as pd
df = pd.read_csv('hamrobazaar_extracted_products.csv')
print(df.info())
print(df.describe())
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
df['Price_NPR'].hist(bins=30, alpha=0.7)
plt.title('Distribution of Asking Prices')
plt.xlabel('Price (NPR)')
plt.ylabel('Frequency')
plt.show()
```

```python
correlation_matrix = df[['original_price', 'Price_NPR', 'used_months', 'storage']].corr()
print(correlation_matrix)
```
